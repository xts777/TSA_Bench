```python
import torch
from chronos import ChronosPipeline, BaseChronosPipeline
from chronos.chronos import ChronosConfig

def load_chronos_model(model_name="amazon/chronos-2", device="cuda"):
    original_init = ChronosConfig.__init__
    def patched_init(self, **kwargs):
        kwargs.pop('input_patch_size', None)
        kwargs.pop('torch_dtype', None)
        original_init(self, **kwargs)
    ChronosConfig.__init__ = patched_init

    print(f"Loading {model_name}...")
    try:
        pipeline = BaseChronosPipeline.from_pretrained(
            model_name,
            device_map=device,
            dtype=torch.float32,
        )
    except Exception as e:
        print(f"Error loading {model_name}: {e}. Fallback to bolt-small.")
        pipeline = ChronosPipeline.from_pretrained(
            "amazon/chronos-bolt-small", 
            device_map=device, 
            torch_dtype=torch.float32
        )
    return pipeline

pipeline = load_chronos_model()
```

```python
for name, module in pipeline.model.named_children():
    print(name, module)
```
